Natural Language Toolkit (NLTK) 라이브러리를 사용한 예제

```
import nltk
nltk.download('wordnet')
```

```
# WordNet 라이브러리를 사용하기 위해 NLTK에서 wordnet 모듈을 불러옴
from nltk.corpus import wordnet

# 'dog'라는 단어와 관련된 모든 synset(동의어 집합)을 가져옴
# Synset: WordNet에서 단어의 여러 의미(sense)를 묶어 놓은 집합
synsets = wordnet.synsets('dog')

# synsets 리스트 중 첫 번째 synset의 정의(definition)를 확인
# synsets[0] → 'dog'의 가장 대표적인 의미(보통 빈도가 높은 의미)
print("dog의 정의:", synsets[0].definition())
```

```
# WordNet에서 'happy'라는 단어의 모든 synset(동의어 집합)을 가져옴
happy_synsets = wordnet.synsets('happy')

# 동의어(lemma) 추출
# 각 synset 안에는 여러 lemma(단어 형태)가 포함되어 있음
# synset.lemmas() → 해당 의미를 나타내는 단어 리스트
# lemma.name() → lemma의 실제 단어 문자열을 가져옴
print("happy의 동의어:", [lemma.name() for synset in happy_synsets for lemma in synset.lemmas()])

# 반의어 추출
# lemma.antonyms() → 해당 lemma의 반의어 리스트를 반환 (비어 있을 수도 있음)
# if lemma.antonyms() → 반의어가 있는 경우만 선택
# lemma.antonyms()[0].name() → 첫 번째 반의어 단어 문자열을 가져옴
print("happy의 반의어:", [lemma.antonyms()[0].name() for synset in happy_synsets for lemma in synset.lemmas() if lemma.antonyms()])
```

```
# 'dog' 단어의 첫 번째 synset 선택 (보통 가장 일반적인 의미: 애완견/가축화된 개)
dog_synset = synsets[0]

# 상위어(hypernym) 확인
# hypernyms() → 해당 synset이 속하는 더 일반적인 개념(상위 개념)들을 반환
hypernyms = dog_synset.hypernyms()
# 각 상위어 synset의 이름(name)을 출력
print("dog의 상위어:", [hypernym.name() for hypernym in hypernyms])

# 하위어(hyponym) 확인
# hyponyms() → 해당 synset에 속하는 더 구체적인 개념(하위 개념)들을 반환
hyponyms = dog_synset.hyponyms()
# 각 하위어 synset의 이름(name)을 출력
print("dog의 하위어:", [hyponym.name() for hyponym in hyponyms])
```

# 2. Lesk 알고리즘
### Lesk 알고리즘을 사용하여 주어진 문장에서 'word'의 가장 적절한 의미(sense)를 찾는 함수
***단어와  문장에 나타난 단어에 대해  Best Sense 추출***

```
# Lesk 알고리즘을 사용하여 주어진 문장에서 'word'의 가장 적절한 의미(sense)를 찾는 함수
def disambiguate(word, sentence, stopwords):

    word_senses = wordnet.synsets(word)
    # WordNet에서 해당 단어(word)의 모든 의미(synset)를 가져옴

    best_sense = word_senses[0]
    # 초기값: 첫 번째 의미를 기본값으로 설정 (WordNet은 보통 가장 빈도가 높은 의미를 첫 번째로 둠)

    max_overlap = 0
    # 최대 겹침 수를 기록하기 위한 변수 (현재까지 가장 많이 겹친 단어 수 저장)

    context = set(word_tokenize(sentence))
    # 입력 문장을 단어 단위로 토큰화한 후, 집합(set)으로 변환
    # 문맥(context) 단어들의 집합 → 비교 대상

    for sense in word_senses:
        # 후보 의미(sense) 하나씩 확인
        signature = tokenized_gloss(sense)
        # 해당 의미의 정의(gloss)와 예문(example)을 토큰화하여 signature(대표 단어 집합) 생성

        overlap = compute_overlap(signature, context, stopwords)
        # signature 단어들과 문맥(context) 단어들 사이의 겹치는 단어 수 계산
        # stopwords(불용어)는 제외하여 의미 있는 단어만 비교

        if overlap > max_overlap:
            # 지금까지 찾은 것보다 더 많이 겹친 경우
            max_overlap = overlap
            best_sense = sense
            # 가장 적절한 의미 후보(best_sense) 갱신

    return best_sense
    # 문맥과 가장 많이 겹친 의미(sense)를 최종 반환
```

***sense의 definition에 대한 모든 token 추출***

```
# 특정 synset(단어의 한 의미)에 대해 정의(definition)와 예문(examples)을 토큰화하여 단어 집합 생성
def tokenized_gloss(sense):
    tokens = set(word_tokenize(sense.definition()))
    # synset의 정의(definition)를 단어 단위로 토큰화하여 집합(set)으로 변환
    # 예: 'dog' → "a domesticated carnivorous mammal" → {"a", "domesticated", "carnivorous", "mammal"}

    for example in sense.examples():
        # 해당 synset이 가진 모든 예문(example)에 대해 반복
        tokens = tokens.union(set(word_tokenize(example)))
        # 예문을 토큰화한 집합과 기존 tokens를 합집합(union)으로 결합

    return tokens
    # 정의와 예문에 등장한 모든 단어들의 집합 반환
```

***겹치는 단어 비교***

```
# signature: synset의 정의(definition)와 예문(example)에서 추출한 단어 집합
# context: 문맥 문장에서 추출한 단어 집합
# stopwords: 불용어 집합 (의미 없는 단어 제거용)

def compute_overlap(signature, context, stopwords):
    gloss = signature.difference(stopwords)
    # signature 집합에서 stopwords를 제거하여 의미 있는 단어만 남김
    # 예: {"the", "dog", "animal"} - {"the"} → {"dog", "animal"}

    return len(gloss.intersection(context))
    # 문맥(context)과 signature(gloss) 간의 교집합 크기(겹치는 단어 수)를 반환
    # 즉, 문맥 속 단어와 synset 정의/예문 속 단어가 얼마나 겹치는지 수치화
```

***Lesk 알고리즘을 실제로 동작시키는 예제***

```
# 입력 문장 정의
sentence = "They eat a meal"

# 문장을 단어 단위로 토큰화하여 문맥(context) 집합 생성
# 예: {"They", "eat", "a", "meal"}
context = set(word_tokenize(sentence))

# 의미 중의성 해소 대상 단어
word = 'eat'

print("🔎 Target Word:", word)
print("📌 Input Sentence:", sentence)
print("="*50)
```

# 3. 그래프 기반(NetworkX + WordNet) WSD 구현 예제

```
# =============================
# Graph-based WSD (Knowledge-based) with WordNet + PageRank
# =============================

# (1) 필요한 라이브러리 임포트
import nltk
from nltk.corpus import wordnet as wn
from nltk import word_tokenize
from nltk.corpus import stopwords as nltk_stopwords
import networkx as nx

# ※ 처음 한 번만 실행(리소스 다운로드)
# nltk.download('wordnet')
# nltk.download('omw-1.4')   # WordNet 유사도 개선(권장)
# nltk.download('punkt')
# nltk.download('stopwords')
```

```
# (2) 유틸 함수들
def get_synsets(word, max_n=3):
    """
    주어진 단어의 synset 리스트를 상위 max_n개만 반환.
    (그래프 크기를 과도하게 키우지 않기 위한 제한)
    """
    return wn.synsets(word)[:max_n]

def wup_sim(s1, s2):
    """
    Wu–Palmer 유사도 계산(동일 품사일 때 잘 동작).
    None이 나오면 0.0으로 처리하여 안정화.
    """
    sim = s1.wup_similarity(s2)
    return float(sim) if sim is not None else 0.0

def build_semantic_graph(target_word, context_words, max_synsets_each=3, sim_threshold=0.1):
    """
    의미 그래프 생성:
    - 노드: (word, synset) 튜플
    - 엣지: synset 간 유사도(Wu–Palmer) > threshold 인 경우, 가중치 = 유사도
    """
    G = nx.Graph()

    # 대상 단어와 문맥 단어들의 synset 후보 수집
    cand_by_word = {}
    cand_by_word[target_word] = get_synsets(target_word, max_synsets_each)
    for w in context_words:
        cand_by_word[w] = get_synsets(w, max_synsets_each)

    # 노드 추가: (word, synset) 형태로 식별하기 쉽게 둠
    for w, syns in cand_by_word.items():
        for s in syns:
            G.add_node((w, s))

    # 엣지 추가: 모든 synset 쌍 간 유사도 계산(과한 계산을 피하려면 품사 매칭만)
    nodes = list(G.nodes())
    for i in range(len(nodes)):
        for j in range(i + 1, len(nodes)):
            (w1, s1), (w2, s2) = nodes[i], nodes[j]
            # 품사가 같을 때만 비교(잡음 감소)
            if s1.pos() == s2.pos():
                sim = wup_sim(s1, s2)
                if sim >= sim_threshold:
                    G.add_edge((w1, s1), (w2, s2), weight=sim)

    return G, cand_by_word

def graph_based_wsd(target_word, sentence):
    """
    그래프 기반 WSD(개념):
    1) 문장을 토큰화 & 불용어 제거 → 문맥 단어 추출
    2) 대상/문맥 단어의 synset 후보로 그래프 구성(엣지=유사도)
    3) Personalized PageRank 실행(문맥 노드에 가중치 부여)
    4) 대상 단어의 후보 synset 중 PageRank 점수 최대를 선택
    """
    # 1) 문맥 준비
    stop_words = set(nltk_stopwords.words('english'))
    tokens = [t.lower() for t in word_tokenize(sentence)]
    # 대상 단어는 남기되, 나머지는 불용어 제거
    context_words = [t for t in tokens if t.isalpha() and t not in stop_words and t != target_word.lower()]

    # 2) 그래프 구성
    G, cand_by_word = build_semantic_graph(
        target_word=target_word.lower(),
        context_words=context_words,
        max_synsets_each=3,    # 각 단어당 최대 3개 synset만 사용(시연용)
        sim_threshold=0.1      # 유사도 임계값(낮추면 엣지 많아짐)
    )

    # 3) Personalized PageRank용 personalization 벡터 생성
    #    아이디어: "문맥 단어들의 후보 노드"에 확률 질량을 실어주면,
    #    랜덤 워크가 이들에서 시작/재시작되며 대상 후보 의미와 연결된 쪽에 점수를 몰아줌.
    personalization = {node: 0.0 for node in G.nodes()}
    context_nodes = []
    for w in context_words:
        for s in cand_by_word.get(w, []):
            node = (w, s)
            if node in personalization:
                personalization[node] = 1.0  # 문맥 노드에 가중치 부여
                context_nodes.append(node)

    # 만약 문맥 노드가 하나도 없으면 균등 분포로 대체
    if not any(personalization.values()):
        personalization = None  # networkx가 기본 균등 분포 사용

    # PageRank 실행 (가중치 사용, damping=0.85 기본)
    pr = nx.pagerank(G, alpha=0.85, personalization=personalization, weight='weight')

    # 4) 대상 단어의 후보 synset 중 PageRank 점수 최대를 선택
    target_candidates = [(w, s) for (w, s) in G.nodes() if w == target_word.lower()]
    if not target_candidates:
        return None, pr, G  # 후보가 없으면 None

    best_node = max(target_candidates, key=lambda n: pr.get(n, 0.0))
    best_synset = best_node[1]
    return best_synset, pr, G
```

```
# =============================
# (3) Demo: Sentence and Target Word
# =============================
sentence = "I sat on the bank of the river"
target = "bank"

best, pr_scores, graph = graph_based_wsd(target, sentence)

# 출력 결과 보기 좋게 정리
print("📌 Input Sentence:", sentence)
print("🔎 Target Word   :", target)
print("=" * 60)

# 후보 의미 출력
cands = wn.synsets(target)[:5]   # 최대 5개까지만 출력
print("➡ Candidate Senses for 'bank':")
for i, s in enumerate(cands):
    print(f"{i+1}. {s.name():<20} | POS={s.pos()} | Definition: {s.definition()}")
print("=" * 60)

# 그래프 기반 선택 결과 출력
if best is not None:
    print("✅ Best Sense Selected (Graph-based WSD):")
    print(f" - Synset name : {best.name()}")
    print(f" - POS         : {best.pos()}")
    print(f" - Definition  : {best.definition()}")
    print("=" * 60)
else:
    print("❌ No best sense found (graph was empty).")

# (선택) PageRank 점수 상위 3개 출력 → 어떤 후보들이 점수를 많이 받았는지 확인
print("📊 Top-ranked Synsets (PageRank scores):")
sorted_pr = sorted(pr_scores.items(), key=lambda x: x[1], reverse=True)
for (w, s), score in sorted_pr[:3]:
    print(f" - {w}: {s.name()}  (score={score:.4f})")
```
