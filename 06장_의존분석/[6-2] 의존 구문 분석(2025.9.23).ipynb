
# 영어 의존 구문 분석

Spacy 모델은 문장을 token들로 구성된 document로 처리한다.

각 token에는 품사, 의존 관계, 개체명 정보 등이 태깅된다.

이 예제에서 출력하는 정보들은 다음과 같다.

- token.text: token 문자열
- token.dep_: token과 token의 지배소 간의 의존 관계 유형
- token.head: 지배소 token

```
# spaCy 라이브러리를 불러오기
import spacy

# 영어용 사전 학습된 소형 모델(en_core_web_sm)을 불러옴
# 이 모델은 토큰화(tokenization), 품사 태깅(POS tagging),
# 개체명 인식(NER), 의존 구문 분석(Dependency Parsing) 등
# 여러 NLP 작업을 동시에 처리할 수 있는 "멀티 태스크 통계 모델"
nlp = spacy.load('en_core_web_sm')
```

```
# 'The fat cat sat on the mat'라는 영어 문장을 nlp 모델에 입력
# nlp는 spaCy에서 불러온 언어 모델(en_core_web_sm)로,
# 입력 문장을 분석하여 토큰화, 품사 태깅, 의존 구문 분석, 개체명 인식 등의 결과를 포함하는
# Doc 객체를 반환
doc = nlp('The fat cat sat on the mat')
```

```
# doc 객체에 포함된 모든 단어(token)를 순서대로 반복
for token in doc:
    # 각 단어(token)에 대해 출력할 정보:
    # token.text : 단어 자체 (예: "cat")
    # token.dep_ : 단어의 의존 구문 역할 (dependency label, 예: "nsubj"=주어)
    # token.head.text : 이 단어가 의존하는 중심 단어(Head)
    print(token.text, token.dep_, token.head.text)
```

의존 구문 분석 결과를 시각화한다.

```
# spaCy의 시각화 모듈 displacy 불러오기
# displacy는 의존 구문 분석 트리나 개체명 인식 결과를 시각적으로 표현하는 도구
from spacy import displacy
```

```
# Jupyter Notebook, Google Colab 등에서 동작
# displacy.render : spaCy의 시각화 함수
# doc : 분석된 문서 객체
# style='dep' : 의존 구문 분석 트리를 그리도록 지정
# jupyter=True : 주피터/코랩 환경에 맞게 HTML로 바로 출력
displacy.render(doc, style='dep', jupyter=True)
```

```
# 영어 문장 입력
doc = nlp('I saw the man with the telescope.')

# 문장을 토큰(단어 단위)별로 순회
for token in doc:
    # token.text : 단어 자체
    # token.dep_ : 단어의 의존 관계 라벨 (예: nsubj=주어, dobj=목적어, prep=전치사)
    # token.head.text : 해당 단어가 의존하는 중심 단어(Head)
    print(token.text, token.dep_, token.head.text)
```

```
displacy.render(doc, style='dep', jupyter=True)
```

```
doc = nlp('Time flies like an arrow.')

for token in doc:
  print(token.text, token.dep_, token.head.text)

displacy.render(doc, style='dep', jupyter=True)
```

## spaCy: 토큰화, 품사, 의존구문, NER, 시각화

```
import spacy
from spacy import displacy

# 중국어 모델 로드 (소형: zh_core_web_sm / 고성능: zh_core_web_trf)
nlp = spacy.load("zh_core_web_sm")

# 예문 1: 기본 SVO + 전치사구
doc1 = nlp("我在学校学习中文。")

# 예문 2: 把(ba) 처치문
doc2 = nlp("张三把苹果吃了。")

# 예문 3: 被(bei) 수동문
doc3 = nlp("书被他买了。")

# 토큰/품사/의존관계 출력
def show_dep(doc):
    print("====", doc.text, "====")
    for tok in doc:
        print(f"{tok.text:6s}  POS={tok.pos_:6s}  DEP={tok.dep_:8s}  HEAD={tok.head.text}")
    print()

show_dep(doc1)
show_dep(doc2)
show_dep(doc3)

# 의존구문 트리 시각화(주피터/코랩)
displacy.render(doc1, style="dep", jupyter=True)
displacy.render(doc2, style="dep", jupyter=True)
displacy.render(doc3, style="dep", jupyter=True)

# 개체명 인식 예시
doc_ner = nlp("马云在杭州创办了阿里巴巴集团。")
for ent in doc_ner.ents:
    print(ent.text, ent.label_)  # 예: PERSON, GPE, ORG 등
displacy.render(doc_ner, style="ent", jupyter=True)
```

## jieba: 중국어 분절(띄어쓰기 부재 해결)

```
import jieba

s = "我喜欢自然语言处理，也喜欢用Python做项目。"
# 기본 분절
print("/".join(jieba.cut(s)))

# 검색용(더 세밀)
print("/".join(jieba.cut_for_search("上海市长江大桥")))
# → “上海/上海市/市长/长江/长江大桥/大桥” 등 다의적 분절 관찰 가능
```

## stanza: 품사/의존구문(UD) 파이프라인

```
import stanza
# 최초 1회: stanza.download("zh")
nlp_stz = stanza.Pipeline("zh")  # 중국어 기본 UD 파이프라인

text = "我在学校学习中文。张三把苹果吃了。书被他买了。"
doc = nlp_stz(text)

for i, sent in enumerate(doc.sentences, 1):
    print(f"=== Sentence {i}: {''.join([w.text for w in sent.words])} ===")
    for w in sent.words:
        # id, text, upos(품사), head(지배소 ID), deprel(의존관계)
        print(f"{w.id:>2} {w.text:<4} UPOS={w.upos:<6} HEAD={w.head:<2} DEPREL={w.deprel}")
    print()
```

```
# '行'의 다의성
doc_poly = nlp("这样行吗？我觉得行。银行也行。")
for tok in doc_poly:
    print(tok.text, tok.pos_, tok.dep_, tok.head.text)

# 전치사구 결합 모호성 (망원경 예제의 중국어판)
doc_amb = nlp("我看见拿着望远镜的人。")
displacy.render(doc_amb, style="dep", jupyter=True)
```

# 한국어 의존구문 분석 예제

```
# 설치 (최초 1회)
# !pip install stanza spacy spacy-stanza

import spacy_stanza
from spacy import displacy

# 1) 한국어 파이프라인을 spaCy 형태로 로드
nlp = spacy_stanza.load_pipeline("ko")

# 2) 예문 입력
doc = nlp("나는 학교에 간다.")

# 3) displacy로 의존 트리 시각화 (Jupyter/Colab 환경)
displacy.render(doc, style="dep", jupyter=True, options={"compact": True, "distance": 110})
```
