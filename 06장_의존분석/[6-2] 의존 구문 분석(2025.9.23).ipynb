
# 영어 의존 구문 분석

Spacy 모델은 문장을 token들로 구성된 document로 처리한다.

각 token에는 품사, 의존 관계, 개체명 정보 등이 태깅된다.

이 예제에서 출력하는 정보들은 다음과 같다.

- token.text: token 문자열
- token.dep_: token과 token의 지배소 간의 의존 관계 유형
- token.head: 지배소 token

```
# spaCy 라이브러리를 불러오기
import spacy

# 영어용 사전 학습된 소형 모델(en_core_web_sm)을 불러옴
# 이 모델은 토큰화(tokenization), 품사 태깅(POS tagging),
# 개체명 인식(NER), 의존 구문 분석(Dependency Parsing) 등
# 여러 NLP 작업을 동시에 처리할 수 있는 "멀티 태스크 통계 모델"
nlp = spacy.load('en_core_web_sm')
```

```
# 'The fat cat sat on the mat'라는 영어 문장을 nlp 모델에 입력
# nlp는 spaCy에서 불러온 언어 모델(en_core_web_sm)로,
# 입력 문장을 분석하여 토큰화, 품사 태깅, 의존 구문 분석, 개체명 인식 등의 결과를 포함하는
# Doc 객체를 반환
doc = nlp('The fat cat sat on the mat')
```

```
# doc 객체에 포함된 모든 단어(token)를 순서대로 반복
for token in doc:
    # 각 단어(token)에 대해 출력할 정보:
    # token.text : 단어 자체 (예: "cat")
    # token.dep_ : 단어의 의존 구문 역할 (dependency label, 예: "nsubj"=주어)
    # token.head.text : 이 단어가 의존하는 중심 단어(Head)
    print(token.text, token.dep_, token.head.text)
```

의존 구문 분석 결과를 시각화한다.

```
# spaCy의 시각화 모듈 displacy 불러오기
# displacy는 의존 구문 분석 트리나 개체명 인식 결과를 시각적으로 표현하는 도구
from spacy import displacy
```

```
# Jupyter Notebook, Google Colab 등에서 동작
# displacy.render : spaCy의 시각화 함수
# doc : 분석된 문서 객체
# style='dep' : 의존 구문 분석 트리를 그리도록 지정
# jupyter=True : 주피터/코랩 환경에 맞게 HTML로 바로 출력
displacy.render(doc, style='dep', jupyter=True)
```

```
# 영어 문장 입력
doc = nlp('I saw the man with the telescope.')

# 문장을 토큰(단어 단위)별로 순회
for token in doc:
    # token.text : 단어 자체
    # token.dep_ : 단어의 의존 관계 라벨 (예: nsubj=주어, dobj=목적어, prep=전치사)
    # token.head.text : 해당 단어가 의존하는 중심 단어(Head)
    print(token.text, token.dep_, token.head.text)
```

```
displacy.render(doc, style='dep', jupyter=True)
```

```
# === 분석할 문장: "Time flies like an arrow." (시간은 화살처럼 날아간다)
doc = nlp('Time flies like an arrow.')

# === 토큰(단어) 단위로 구문 분석 결과 출력 ===
for token in doc:
    # token.text : 단어 자체
    # token.dep_ : 단어의 문법적 역할 (예: 주어, 목적어, 수식어 등)
    # token.head.text : 이 단어가 의존하는 중심 단어(head)
    print(token.text, token.dep_, token.head.text)

# === 구문 트리 시각화 ===
# style='dep' → 의존 구문(Dependency Parsing) 결과를 그림으로 표시
# jupyter=True → Jupyter Notebook 환경에서 바로 출력
displacy.render(doc, style='dep', jupyter=True)
```

## spaCy: 토큰화, 품사, 의존구문, NER, 시각화

```
import spacy
from spacy import displacy

# 중국어 모델 로드 (소형: zh_core_web_sm / 고성능: zh_core_web_trf)
nlp = spacy.load("zh_core_web_sm")

# 예문 1: 기본 SVO + 전치사구
doc1 = nlp("我在学校学习中文。")

# 예문 2: 把(ba) 처치문
doc2 = nlp("张三把苹果吃了。")

# 예문 3: 被(bei) 수동문
doc3 = nlp("书被他买了。")

# 토큰/품사/의존관계 출력
def show_dep(doc):
    print("====", doc.text, "====")
    for tok in doc:
        print(f"{tok.text:6s}  POS={tok.pos_:6s}  DEP={tok.dep_:8s}  HEAD={tok.head.text}")
    print()

show_dep(doc1)
show_dep(doc2)
show_dep(doc3)

# 의존구문 트리 시각화(주피터/코랩)
displacy.render(doc1, style="dep", jupyter=True)
displacy.render(doc2, style="dep", jupyter=True)
displacy.render(doc3, style="dep", jupyter=True)

# 개체명 인식 예시
doc_ner = nlp("马云在杭州创办了阿里巴巴集团。")
for ent in doc_ner.ents:
    print(ent.text, ent.label_)  # 예: PERSON, GPE, ORG 등
displacy.render(doc_ner, style="ent", jupyter=True)
```

## jieba: 중국어 분절(띄어쓰기 부재 해결)

jieba.cut() : 일반적인 단어 분절 (보통 텍스트 처리에 사용)

jieba.cut_for_search() : 검색용 세밀한 분절 (검색 엔진에서 더 많은 후보어를 잡기 위해 사용)

예시 "上海市长江大桥" → "上海市", "长江大桥", "市长", "大桥" 등 여러 방식으로 분리

```
import jieba

# === 중국어 문장 ===
s = "我喜欢自然语言处理，也喜欢用Python做项目。"
# 의미: "나는 자연어처리를 좋아하고, Python으로 프로젝트 하는 것도 좋아한다."

# === 1) 기본 분절 ===
# jieba.cut(s) → 문장을 일반적인 단어 단위로 분리
# "/".join(...) → 분리된 단어들을 "/" 기호로 연결해 출력
print("/".join(jieba.cut(s)))
# 예) 我/喜欢/自然语言处理/，/也/喜欢/用/Python/做/项目/。

# === 2) 검색용 분절 ===
# jieba.cut_for_search(...) → 검색 엔진 최적화를 위해
# 좀 더 세밀하게(중복 포함) 단어를 쪼갬
print("/".join(jieba.cut_for_search("上海市长江大桥")))
# 출력 예시: 上海/上海市/市长/长江/长江大桥/大桥
# → '上海市长江大桥' 같은 긴 단어를
#    여러 가지 가능한 작은 단위로도 동시에 쪼개어 줌
#    (검색 시 다의적 분절 효과를 얻을 수 있음)
```

## stanza: 품사/의존구문(UD) 파이프라인

중국어 문법 구조(주어-술어-목적어 / 처치문 / 피동문)가
의존구문분석으로 어떻게 표현되는지 확인하는 예제

```
import stanza
# === 중국어 Stanza 파이프라인 준비 ===
# 최초 1회만 실행 (모델 다운로드)
# stanza.download("zh")

# "zh" = 중국어 모델, Universal Dependencies(UD) 기반 파이프라인 사용
nlp_stz = stanza.Pipeline("zh")

# === 입력 문장 ===
text = "我在学校学习中文。张三把苹果吃了。书被他买了。"
# 의미:
# 1) 나는 학교에서 중국어를 공부한다.
# 2) 장삼은 사과를 먹었다.
# 3) 책은 그가 샀다. (피동문)

# === 분석 실행 ===
doc = nlp_stz(text)

# === 문장 단위로 결과 출력 ===
for i, sent in enumerate(doc.sentences, 1):
    # 문장 번호 + 원문 출력
    print(f"=== Sentence {i}: {''.join([w.text for w in sent.words])} ===")
    for w in sent.words:
        # w.id      : 단어 번호 (문장 내 위치)
        # w.text    : 단어 자체
        # w.upos    : 품사 (UD 표준, e.g. NOUN, VERB, PRON 등)
        # w.head    : 의존하는 단어의 ID (head word)
        # w.deprel  : 의존관계 (nsubj=주어, obj=목적어, case=격조사 등)
        print(f"{w.id:>2} {w.text:<4} UPOS={w.upos:<6} HEAD={w.head:<2} DEPREL={w.deprel}")
    print()
```

중국어의 의미 중의성(polysemy)과 구문 중의성(syntactic ambiguity)을 보여주는 예제

展示中文的语义多义性（polysemy）和句法歧义（syntactic ambiguity）的例子

```
# === '行'의 다의성(Polysemy) ===
# "行" 은 문맥에 따라 '되다(ok)', '가능하다', '은행' 등으로 해석됨
doc_poly = nlp("这样行吗？我觉得行。银行也行。")
for tok in doc_poly:
    # tok.text  : 단어 자체
    # tok.pos_  : 품사 (동사/명사 등)
    # tok.dep_  : 문장에서의 문법적 역할
    # tok.head.text : 의존하는 중심 단어(head)
    print(tok.text, tok.pos_, tok.dep_, tok.head.text)

# 출력 예시 해석
# "这样行吗？" → '行' 은 동사(괜찮다/가능하다)로 해석
# "我觉得行。" → '行' 은 '좋다/괜찮다' 의미로 동사
# "银行也行。" → '行' 은 '은행(银行)'의 일부로 명사적 의미
# → 같은 글자 '行' 이 문맥마다 다른 의미·품사로 인식됨

# === 전치사구 결합 모호성 (망원경 예제) ===
# "我看见拿着望远镜的人。" → "나는 망원경을 든 사람을 보았다."
# 그러나 해석은 두 가지 가능:
#   1) 내가 망원경을 사용해 사람을 보았다.
#   2) 망원경을 든 사람을 보았다.
doc_amb = nlp("我看见拿着望远镜的人。")

# displaCy로 의존구문 트리 시각화
# style="dep" → 의존관계 시각화
# jupyter=True → Jupyter Notebook에서 실행 시 그림으로 표시
displacy.render(doc_amb, style="dep", jupyter=True)
```

# 한국어 의존구문 분석 예제

spacy_stanza.load_pipeline("ko") → 한국어 문법 분석기 로드

doc = nlp("...") → 입력 문장 분석

displacy.render(...) → 문장 구조(주어, 목적어, 동사 관계)를 트리 그림으로 보여줌

```
# === 설치 (최초 1회만 실행) ===
# !pip install stanza spacy spacy-stanza
# → 한국어 모델을 사용하려면 stanza와 spacy-stanza 패키지가 필요함

!pip install spacy-stanza
```

```
import spacy_stanza
from spacy import displacy

# 1) 한국어 파이프라인을 spaCy 형태로 로드
# "ko" = 한국어 모델
# spacy_stanza.load_pipeline("ko") → Stanza의 한국어 모델을 불러와서 spaCy 스타일로 사용 가능
nlp = spacy_stanza.load_pipeline("ko")
```

```
# 2) 분석할 문장 입력
doc = nlp("나는 학교에 간다.")
# 의미: "I go to school."

# === displacy로 의존 트리 시각화 ===
# style="dep" : 의존 관계 분석 결과를 그림으로 표현
# jupyter=True : Jupyter/Colab 환경에서 바로 시각화
# options : compact=True → 간단히, distance=110 → 단어 간격 넓게
displacy.render(doc, style="dep", jupyter=True, options={"compact": True, "distance": 110})
```

```
# === 한국어 문장 의존구문 분석 ===
# 문장: "철수가 영희에게 편지를 보냈다."
# 의미: 철수가 (주어) / 영희에게 (간접목적어) / 편지를 (직접목적어) / 보냈다 (서술어)

doc = nlp("철수가 영희에게 편지를 보냈다.")

# === displacy로 의존 트리 시각화 ===
displacy.render(doc, style="dep", jupyter=True,
                options={"compact": True, "distance": 110})
```

```
# === 한국어 문장 의존구문 분석 ===
# 문장: "그 책이 학생들에게 읽혔다."
# 의미: "That book was read by the students." (피동문)

doc = nlp("그 책이 학생들에게 읽혔다.")

# === displacy로 의존 트리 시각화 ===
displacy.render(doc, style="dep", jupyter=True,
                options={"compact": True, "distance": 110})
```
