NLTK로 토큰화, 품사 태깅, 개체명 인식을 수행하기 위해 필요한 데이터 파일들을 미리 다운로드

```python
# -*- coding:utf-8 -*-
# 파이썬 소스 코드가 UTF-8 문자 인코딩을 사용한다는 선언 (한글 등 처리 가능)

import nltk  # 자연어처리 라이브러리 NLTK 불러오기

# 아래는 NLTK에서 필요한 리소스(데이터/모델) 다운로드하는 부분입니다.
# 처음 한 번만 실행하면 됩니다.

nltk.download('punkt')
# 문장을 단어로 나누는 토큰화(Tokenization) 도구

nltk.download('punkt_tab')
# 토큰화 관련 추가 리소스 (NLTK 최신 버전에서 요구할 수 있음)

nltk.download('words')
# 영어 단어 리스트 (사전처럼 활용)

nltk.download('averaged_perceptron_tagger')
# 품사 태깅(POS Tagging)을 위한 모델 (단어의 문법적 역할 예측)

nltk.download('averaged_perceptron_tagger_eng')
# 영어 전용 품사 태깅 모델 (추가 리소스)

nltk.download('maxent_ne_chunker')
# 개체명 인식(NER)을 위한 MaxEnt 모델 (사람, 장소, 기관명 등 추출)

nltk.download('maxent_ne_chunker_tab')
# 개체명 인식 관련 추가 리소스 (NLTK 최신 버전에서 요구될 수 있음)
```

```python
# sentence = "Prime Minister Boris Johnson had previously said the UK would leave by 31 October."
sentence = "Barack Obama was born in Hawaii."
```

```python
# 문장을 단어 단위로 쪼개기 (tokenize)
# 즉, 한 문장을 "단어 목록"으로 나누어 줌
tokens = nltk.word_tokenize(sentence)

print(tokens)
```

```python
# 단어에 품사(Part-of-Speech, POS) 태그 붙이기
# 예: 명사(NN), 동사(VB), 형용사(JJ) 등
tagged = nltk.pos_tag(tokens)

print(tagged)
```

```python
# 개체명 인식 (Named Entity Recognition, NER)
# 문장에서 인물(Person), 장소(Location), 기관(Organization) 등의 개체를 찾아 트리 구조로 표시
entities = nltk.chunk.ne_chunk(tagged)

print(entities)
```

### spaCy 패키지에서 NER(Named Entity Recognition)을 지원하는 언어

| 언어                 | 특징 / 비고                                                                                           |
| ------------------ | ------------------------------------------------------------------------------------------------- |
| 영어 (English)       | 가장 완전한 지원. 여러 크기의 사전학습 모델(en_core_web_sm, md, lg, trf 등) 제공                      |
| 중국어 (Chinese / zh) | 중국어 모델이 제공되며, 단어 분절(segmentation) 옵션(jieba, pkuseg, char) 조정 가능                    |
| 일본어 (Japanese)     | sudachipy 기반 단어 분절 및 품사 태깅, NER 포함 가능                                              |
| 한국어 (Korean)       | MeCab 기반 토크나이저 지원. 다만 한국어에 대해 완전한 NER 모델이 기본 제공되는지는 제한적임 (모델/리소스 추가 필요)            |
| 다국어 / 멀티언어 (xx)    | 여러 언어를 아우르는 멀티언어 NER 파이프라인을 spaCy에서 설정 가능. “xx” 언어 클래스를 사용해 여러 언어 통합 모델을 구성할 수 있음  |


```python
import spacy

# 1) 영어용 사전학습 모델 불러오기
# "en_core_web_sm" → 영어 소형 모델 (토큰화, 품사 태깅, 개체명 인식 기능 포함)
nlp = spacy.load("en_core_web_sm")

# 2) 분석할 문장 입력
text = "Apple is looking at buying U.K. startup for $1 billion"

# 3) 문장을 NLP 파이프라인에 통과시켜 처리
doc = nlp(text)

# 4) 문장에서 인식된 개체명(Entities) 출력
# ent.text  → 실제 단어(문자열)
# ent.label_ → 개체 유형(Person, Org, GPE, Money 등)
for ent in doc.ents:
    print(ent.text, ent.label_)

# 예상 출력:
# → Apple = 기관(Organization),
#    U.K. = 지명(Geo-Political Entity),
#    $1 billion = 금액(Money)
```

```python
# 대구대학교는 1956년 고(故) 이영식 박사에 의해 설립되었습니다.
# 한국 전쟁 이후, 전쟁으로 인해 상처받은 사람들에게 교육의 기회를 제공하고자 하는 취지에서 시작되었습니다.

# 여러 줄 텍스트 입력 (대구대학교 설립 관련 예시)
text = """Daegu University was founded in 1956 by Dr. Lee Young-Sik.
After the Korean War, it was established with the purpose of providing educational opportunities to those affected by the war.
"""

# 문장을 NLP 파이프라인에 통과
doc = nlp(text)

# 문장에서 추출된 개체명 출력
for ent in doc.ents:
    print(ent.text, ent.label_)

# 예상 출력 예시:
# Daegu University ORG        → 기관명 (Organization)
# 1956 DATE                   → 연도 (날짜/시간)
# Lee Young-Sik PERSON        → 인명 (Person)
# the Korean War EVENT        → 사건(Event)
```

```python
import spacy

# 1) 중국어용 사전학습 모델 불러오기
#    사전에 설치 필요: python -m spacy download zh_core_web_sm
nlp = spacy.load("zh_core_web_sm")

# 2) 분석할 중국어 문장
text = "我昨天在上海参加了复旦大学的人工智能会议。"
# 번역: 나는 어제 상하이에서 푸단대학교의 인공지능 회의에 참가했다.

# 3) 문장을 NLP 파이프라인에 통과
doc = nlp(text)

# 4) 인식된 개체명(Entities) 출력
for ent in doc.ents:
    print(ent.text, ent.label_)

# 예상 출력 (실제 결과는 모델에 따라 다를 수 있음):
# 昨天 DATE
# 上海 GPE         → 지명 (Location)
# 复旦大学 ORG     → 기관명 (Organization)
```
